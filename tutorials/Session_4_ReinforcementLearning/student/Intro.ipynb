# Intro

## Overview

This session is an intuitive tour of **reinforcement learning (RL)**: how an agent can learn what to do from trial, error, and reward alone, without being told the correct answer.
It follows the same three-part arc as the lecture:

1. **Foundations.** The vocabulary of Markov decision processes, the value function, the Bellman equation, and the explore–exploit dilemma of the multi-armed bandit.
2. **Scaling up: deep RL.** Action values $Q(s,a)$, Q-learning, why plugging in a neural network (DQN) is hard, and actor–critic methods.
3. **RL in the brain.** The temporal-difference (TD) error as the dopamine reward prediction error, and the distinction between model-free (habitual) and model-based (planning) control.

The two hands-on tutorials mirror Part 1 and Part 3:

- **Tutorial 1** builds ε-greedy and softmax agents for a multi-armed bandit.
- **Tutorial 2** reproduces the dopamine recordings of Schultz, Dayan & Montague (1997) with a TD(0) learner, then contrasts model-free and model-based values after reward devaluation.

This session was originally planned as two sessions (RL foundations and RL in the brain) and condensed into one.

### Lecture at a glance

**Return and value.** Reward is what you get now; value is what a situation is worth in the long run.

$$G_t = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + \cdots, \qquad v_\pi(s) = \mathbb{E}_\pi\left[G_t \mid S_t = s\right].$$

**Bellman equation.** The value of a state is the immediate reward plus the discounted value of where you land next:

$$v_\pi(s) = \sum_a \pi(a\mid s) \sum_{s',r} p(s',r\mid s,a)\left[r + \gamma v_\pi(s')\right].$$

**Exploration.** An ε-greedy agent picks the best-looking action with probability $1-\varepsilon$ and a random action otherwise. A softmax agent picks actions with probability

$$\pi(a\mid s) = \frac{e^{Q(s,a)/\tau}}{\sum_b e^{Q(s,b)/\tau}},$$

where a large temperature $\tau$ is more random and a small $\tau$ is more greedy.

**Q-learning (off-policy TD control).**

$$Q(S,A) \leftarrow Q(S,A) + \alpha\left[R + \gamma \max_a Q(S',a) - Q(S,A)\right].$$

**Deep RL.** Replacing the table with a network $\hat q(s,a;\mathbf{w})$ breaks two assumptions of supervised learning: samples are correlated in time and the target moves as the network learns. DQN counters these with an experience-replay buffer and a slowly updated target network.

**Actor–critic.** The actor is the policy; the critic estimates value and tells the actor whether an action turned out better or worse than expected through the TD error

$$\delta_t = R_{t+1} + \gamma V(S_{t+1}) - V(S_t).$$

**Dopamine.** Midbrain dopamine neurons fire in proportion to $\delta_t$: they burst for unexpected rewards, the burst moves to a predictive cue with learning, and they dip below baseline when an expected reward is withheld.

**Model-free vs. model-based.** A model-free agent caches $Q(s,a)$ from experience. A model-based agent learns the transition model $T(s'\mid s,a)$ and reward $R(s,a)$ and plans with them, so it can re-plan immediately when the world changes.
