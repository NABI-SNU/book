# Introduction to Models

## Overview

This session introduces **models as tools for linking experimental data to hypotheses**. We will focus on regression-style models because they make the modeling workflow explicit: define inputs and outputs, choose a model family, fit parameters, and evaluate how well the model generalizes.

The tutorials build up the core ideas step by step using executable examples.

### Tutorial Structure

1. Tutorial 1 introduces multiple linear regression and polynomial regression. You will learn how design matrices represent model assumptions and how ordinary least squares estimates parameters.
2. Tutorial 2 focuses on model selection through the bias-variance trade-off. You will compare train and test error to see why training error alone is not enough.
3. Tutorial 3 introduces generalized linear models (GLMs). You will see how link functions and likelihoods let us model non-Gaussian data such as spike counts.

By the end of this session, you should be able to explain what a model assumes, fit its parameters, and compare candidate models using held-out data or an appropriate likelihood.

**Exercise 1 — Derivation of the Least-Squares Estimator**

Consider a linear model

$$
\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\epsilon},
$$

where  
- $\mathbf{X} \in \mathbb{R}^{n \times p}$ is called the design matrix  
- $\mathbf{y} \in \mathbb{R}^n$ is called the response vector  
- $\boldsymbol{\beta} \in \mathbb{R}^p$ are called the parameters  

We define the least-squares objective:

$$
L(\boldsymbol{\beta}) = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2
$$

Prove that $L(\boldsymbol{\beta})$ is minimized when $\boldsymbol{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y}$. (Hint: take the derivative of $L(\boldsymbol{\beta})$ with respect to $\boldsymbol{\beta}$ and set it to zero.)

## Exercise 2 — Computing the Least-Squares Estimate from Data

We are given the following dataset:

| $x$ | 1 | 2 | 3 | 4 | 5 |
|-----|---|---|---|---|---|
| $y$ | 3.0 | 5.1 | 6.9 | 9.2 | 10.8 |

Assume a linear model of the form:

$$
y = \beta_0 + \beta_1 x
$$

### Tasks

1. Construct the design matrix and response vector in `numpy`. They should look like this:

$$
\mathbf{X} =
\begin{bmatrix}
1 & 1 \\
1 & 2 \\
1 & 3 \\
1 & 4 \\
1 & 5
\end{bmatrix}, \qquad
\mathbf{y} =
\begin{bmatrix}
3.0 \\
5.1 \\
6.9 \\
9.2 \\
10.8
\end{bmatrix}
$$

2. Solve for the least-squares estimate:

$$
\hat{\boldsymbol{\beta}}
=
(\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y}
$$
