## Overview

Two networks can reach the same held-out accuracy while their hidden units carry no obvious labels and use entirely different bases. This session asks **what is represented, and in what form?**

Three method families answer three different questions:

| | RSA | CCGP | DSA |
|---|---|---|---|
| **Question** | Which conditions are similar or dissimilar? Do two models share the same static geometry? | Does one linear readout transfer to unseen conditions? Which geometric structure supports that transfer? | Do the state transitions of hidden states follow similar rules? Are the dynamics equivalent up to coordinate alignment? |
| **Requires time?** | no | no | yes |
| **Typical output** | RDM similarity | transfer accuracy | aligned operator distance |

By the end of the session you will be able to:

- summarize a representation with a representational dissimilarity matrix (RSA);
- distinguish decodability from cross-condition generalization;
- characterize abstraction with CCGP and parallel coding directions;
- relate abstraction to shattering dimensionality;
- trace DSA from delay coordinates to aligned evolution operators;
- choose among RSA, CCGP, and DSA based on the claim you want to make.

### Lecture at a glance

**A representation is a set of response vectors.** For sample $n$ with input $x_n$ and condition label $y_n$, at layer $\ell$:

$$\mathbf{z}^{(\ell)}_n = f^{(\ell)}(x_n) \in \mathbb{R}^{p_\ell}, \qquad \boldsymbol{\mu}^{(\ell)}_c = \frac{1}{N_c}\sum_{y_n = c} \mathbf{z}^{(\ell)}_n.$$

**Coordinates describe a map; distances describe its geometry.** Pairwise distances are unchanged by rotations and reflections of the coordinate basis, so they summarize a representation independently of which units carry it.

**RSA.** Compare condition means with a dissimilarity such as

$$d^{(\ell)}_{ij} = \lVert \boldsymbol{\mu}^{(\ell)}_i - \boldsymbol{\mu}^{(\ell)}_j \rVert_2 \quad\text{or}\quad d^{(\ell)}_{ij} = 1 - \operatorname{corr}(\boldsymbol{\mu}^{(\ell)}_i, \boldsymbol{\mu}^{(\ell)}_j),$$

and arrange them in a $K\times K$ RDM $D^{(\ell)}$ with $K(K-1)/2$ unique entries. RSA compares relations, not units: networks and brain regions with different feature counts can still be compared through their RDMs.

**Abstraction.** A binary variable is represented abstractly when a linear rule learned from some conditions generalizes to new conditions on both sides of that variable. A balanced dichotomy splits the $K$ conditions into two equal groups $C = C_+ \cup C_-$; with eight conditions there are $\tfrac12\binom{8}{4} = 35$ of them.

**CCGP** withholds *conditions*, not merely trials. For one dichotomy, train a linear classifier on some conditions from both sides and test it on conditions it never saw, averaging over admissible splits $S$:

$$\mathrm{CCGP} = \frac{1}{|S|}\sum_{s\in S}\mathrm{Acc}\left(h_s;\, C\setminus C^{\text{train}}_s\right).$$

**Parallelism score (PS).** For paired conditions, the coding direction $\mathbf{v}_j = (\boldsymbol{\mu}_{+,j} - \boldsymbol{\mu}_{-,j}) / \lVert \boldsymbol{\mu}_{+,j} - \boldsymbol{\mu}_{-,j} \rVert_2$; PS averages pairwise cosines between these directions. High PS means aligned condition means, but CCGP also depends on spacing and trial-level noise.

**Static geometry can omit the transition rule.** The same cloud of states can be visited by systems with different dynamics.

**DSA** asks whether the learned evolution operators of two systems $\dot{x} = f(x,t)$ and $\dot{y} = g(y,t)$ agree after a change of coordinates. Delay coordinates $\mathbf{h}_t = [\mathbf{x}_t^\top\ \mathbf{x}_{t-\tau}^\top\ \cdots\ \mathbf{x}_{t-(p-1)\tau}^\top]^\top$ recover state carried by history; fitted operators $A_x, A_y$ are then aligned by conjugation rather than by moving point clouds:

$$d_{\text{shape}}(X,Y) = \min_{C\in O(r)} \lVert X - CY \rVert_F \qquad\text{vs.}\qquad d_{\text{DSA}}(A_x, A_y) = \min_{C\in O(r)} \lVert A_x - C A_y C^{-1} \rVert_F.$$

A DSA score depends on the sampling, embedding, rank, and quality of the fitted model. **Choose the method from the claim, not from the data format alone.**
