# Intro

## Overview

This session focuses on **marginalization**, **Gaussian models**, and the link between **priors** and **penalized regression** (Ridge and Lasso).

- **Tutorial 1 (student worksheet):** SymPy and Matplotlib exercises on marginalizing a **bivariate Gaussian** and interpreting a **mixture of Gaussians** as a marginal over a discrete latent variable. **Instructor** notebooks contain full worked solutions.
- **Tutorial 2 (student worksheet):** Exercises on the **Gaussian linear likelihood** (OLS), **Gaussian priors** and **Ridge (MAP)**, and **Laplace priors** and **Lasso (MAP)**, with SymPy and penalty-shape plots. Solutions are in the **instructor** notebook.

Bayesian thinking still underlies how we combine likelihoods with priors—here you will see that connection concretely in regression and in marginal distributions.
