# Intro

## Overview

This session introduces **dimension reduction as a practical tool for understanding high-dimensional data**. We will focus on PCA, a linear method that is fast and interpretable, and UMAP, a nonlinear method that is useful for exploratory two-dimensional visualization.

The tutorials now move from PCA fundamentals to visualization with MNIST.

### Tutorial Structure

1. Tutorial 1 introduces PCA foundations: covariance, projections, eigenvectors, eigenvalues, and PCA scores.
2. Tutorial 2 compares PCA and UMAP as simple ways to visualize MNIST in two dimensions.

By the end of this session, you should be able to explain what PCA preserves, compute PCA scores from a covariance matrix, and use PCA or UMAP to make a low-dimensional visualization of high-dimensional data.


## Warm-Up Questions

A set of warm-up questions is provided below. It is not mandatory to answer them, but we encourage you to think through them before starting the tutorials.

Let data points be $x_1, \dots, x_n \in \mathbb{R}^d$, and assume they are mean-centered, that is their sum is the $0$ vector in $\mathbb{R}^d$.

---

**Q1. Projection and Variance in $\mathbb{R}^d$**

Let $w \in \mathbb{R}^d$ with $\|w\| = 1$. Define the projection:
$$
z_i = w^\top x_i
$$

Show that:
$$
\mathrm{Var}(z)
= w^\top \left( \frac{1}{n} \sum_{i=1}^n x_i x_i^\top \right) w
$$

Define the covariance matrix:
$$
S = \frac{1}{n} \sum_{i=1}^n x_i x_i^\top
$$

Conclude:
$$
\mathrm{Var}(z) = w^\top S w
$$

---

**Q2. Variance Maximization ⇒ Eigenproblem**

Solve:
$$
\max_{\|w\|=1} w^\top S w
$$

From the Lagrangian, prove that maxima of $w^\top S w$ are given by the eigenvectors of $S$. Furthermore, show that the maximum value of $w^\top S w$ is the largest eigenvalue of $S$, that is 
$$
\boxed{
w^* = v_1, \quad \max w^\top S w = \lambda_1
}
$$
where $v_1, \dots, v_d$ are the eigenvectors of $S$ and $\lambda_1, \dots, \lambda_d$ are the corresponding eigenvalues.
