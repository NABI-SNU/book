**Question 1: Why non-linearity?**

Define a linear classifier as follows:
$$
\mathbf{y} = \mathbf{W}\mathbf{x} + \mathbf{b}
$$
where $\mathbf{x} \in \mathbb{R}^n$ is the input, $\mathbf{y} \in \mathbb{R}^m$ is the output, $\mathbf{W} \in \mathbb{R}^{m \times n}$ is the weight matrix, and $\mathbf{b} \in \mathbb{R}^m$ is the bias vector. Meanwhile, a non-linear classifier has the form:
$$
\mathbf{y} = f(\mathbf{W}\mathbf{x} + \mathbf{b}),
$$
where $f$ is a non-linear function. Linear classifiers are generally cost-efficient to compute. Why then use non-linear functions? 

**Question 1.1. Compositions of linear functions remain linear**

Assume we have two linear classifiers, where the output of the first classifier is used as the input of the second classifier:
$$
\begin{aligned}
\mathbf{h} &= \mathbf{W}_1\mathbf{x} + \mathbf{b}_1, \\
\mathbf{y} &= \mathbf{W}_2\mathbf{h} + \mathbf{b}_2.
\end{aligned}
$$
Show that when $\mathbf{h}$ is substituted into $\mathbf{y}$, the model remains a linear classifier. Argue that any finite composition of linear classifiers reduces to a single linear classifier.

**Question 1.2. Linear functions cannot compute XOR**

Consider the XOR function on binary inputs:
$$
\begin{aligned}
(0,0) &\mapsto 0, & (0,1) &\mapsto 1, \\
(1,0) &\mapsto 1, & (1,1) &\mapsto 0.
\end{aligned}
$$
